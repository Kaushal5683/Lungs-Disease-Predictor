```python
import os
import numpy as np
import pandas as pd
import cv2
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf
from tensorflow.keras.layers import Dense, GlobalAveragePooling2D, Dropout, Input, BatchNormalization, Flatten
from tensorflow.keras.models import Model, load_model
from tensorflow.keras.optimizers import AdamW
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.applications.densenet import preprocess_input
from tensorflow.keras.callbacks import ReduceLROnPlateau, EarlyStopping, LearningRateScheduler
from sklearn.metrics import classification_report
import time
import kagglehub

dataset_path = kagglehub.dataset_download("omkarmanohardalvi/lungs-disease-dataset-4-types")
print(f"Dataset downloaded to: {dataset_path}")

dataset_dir = None
for root, dirs, files in os.walk(dataset_path):
    if "train" in dirs and "val" in dirs and "test" in dirs:
        dataset_dir = root
        break

if dataset_dir is None:
    raise FileNotFoundError("Could not locate the dataset directory. Please check the downloaded files.")

print(f"Dataset directory found: {dataset_dir}")

def count_images_in_directory(directory):
    total_images = 0
    for subdir in os.listdir(directory):
        subdir_path = os.path.join(directory, subdir)
        if os.path.isdir(subdir_path):
            total_images += len([f for f in os.listdir(subdir_path) if f.endswith(('jpg', 'jpeg', 'png'))])
    return total_images

total_train_images = count_images_in_directory(os.path.join(dataset_dir, 'train'))
total_val_images = count_images_in_directory(os.path.join(dataset_dir, 'val'))
total_test_images = count_images_in_directory(os.path.join(dataset_dir, 'test'))

total_images = total_train_images + total_val_images + total_test_images
print("Total train images:", total_train_images)
print("Total validation images:", total_val_images)
print("Total test images:", total_test_images)
print("Total images in the dataset:", total_images)

def visualize_sample_images():
    image_dir = os.path.join(dataset_dir, "train")
    classes = os.listdir(image_dir)
    num_classes = len(classes)
    images_per_class = 5  # Number of images to display per class

    plt.figure(figsize=(20, 10))

    for i, class_name in enumerate(classes):
        class_path = os.path.join(image_dir, class_name)
        image_files = [f for f in os.listdir(class_path) if os.path.isfile(os.path.join(class_path, f))][:images_per_class]

        for j, image_file in enumerate(image_files):
            image_path = os.path.join(class_path, image_file)
            img = cv2.imread(image_path)
            if img is not None:  # Check if the image was loaded successfully
                img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
                plt.subplot(num_classes, images_per_class, i * images_per_class + j + 1)
                plt.imshow(img)
                plt.title(f"{class_name}")
                plt.axis('off')
            else:
                print(f"Error loading image: {image_path}")

    plt.tight_layout()
    plt.show()

visualize_sample_images()

train_dir = os.path.join(dataset_dir, 'train')
contents = os.listdir(train_dir)
num_of_dirs = len([name for name in contents if os.path.isdir(os.path.join(train_dir, name))])

print("Contents of the train directory:")
for item in contents:
    print(item)

print(f"\nNumber of class directories: {num_of_dirs}")

batch_size = 32
num_epochs = 50
image_size = (224, 224)
num_classes = num_of_dirs  # Based on actual directories found

def build_model():
    image_input = Input(shape=(image_size[0], image_size[1], 3))

    mobilenetv2_model = MobileNetV2(weights='imagenet', include_top=False, input_shape=(image_size[0], image_size[1], 3))

    x = GlobalAveragePooling2D()(mobilenetv2_model.output)
    x = Dense(256, activation='relu', kernel_initializer="he_normal")(x)
    x = Dropout(0.5)(x)
    class_outputs = Dense(num_classes, activation='softmax')(x)

    model = Model(inputs=mobilenetv2_model.input, outputs=class_outputs)

    optimizer = AdamW(learning_rate=0.001)
    model.compile(loss='categorical_crossentropy', optimizer=optimizer, metrics=['accuracy'])

    return model

model = build_model()
model.summary()

def create_data_generators():
    train_datagen = ImageDataGenerator(
        preprocessing_function=preprocess_input,
        rotation_range=20,
        width_shift_range=0.2,
        height_shift_range=0.2,
        shear_range=0.2,
        zoom_range=0.2,
        horizontal_flip=True
    )

    val_datagen = ImageDataGenerator(
        preprocessing_function=preprocess_input
    )

    test_datagen = ImageDataGenerator(
        preprocessing_function=preprocess_input
    )

    train_dataset = train_datagen.flow_from_directory(
        os.path.join(dataset_dir, 'train'),
        target_size=image_size,
        batch_size=batch_size,
        class_mode='categorical',
        shuffle=True
    )

    val_dataset = val_datagen.flow_from_directory(
        os.path.join(dataset_dir, 'val'),
        target_size=image_size,
        batch_size=batch_size,
        class_mode='categorical',
        shuffle=False
    )

    test_dataset = test_datagen.flow_from_directory(
        os.path.join(dataset_dir, 'test'),
        target_size=image_size,
        batch_size=batch_size,
        class_mode='categorical',
        shuffle=False
    )

    return train_dataset, val_dataset, test_dataset

train_dataset, val_dataset, test_dataset = create_data_generators()

reduce_lr = ReduceLROnPlateau(
    monitor='val_loss',
    factor=0.5,
    patience=5,
    min_lr=1e-6,
    verbose=1
)

early_stopping = EarlyStopping(
    monitor='val_loss',
    patience=10,
    restore_best_weights=True
)

print("Starting model training...")
history = model.fit(
    train_dataset,
    epochs=num_epochs,
    validation_data=val_dataset,
    callbacks=[reduce_lr, early_stopping]
)

model.save("lung_disease_model.keras")
print("Model saved as 'lung_disease_model.keras'")

print("Evaluating the model on test data...")
start_time = time.time()
evaluation_result = model.evaluate(test_dataset)
evaluation_time = time.time() - start_time
print(f"Time taken to evaluate the model: {evaluation_time:.2f} seconds")

print(f"Test Loss: {evaluation_result[0]:.4f}")
print(f"Test Accuracy: {evaluation_result[1]:.4f}")

y_pred = model.predict(test_dataset)
y_pred_bool = np.argmax(y_pred, axis=1)
true_labels = test_dataset.labels

print("Classification Report:")
print(classification_report(true_labels, y_pred_bool, digits=4))

np.savetxt("y_true.txt", true_labels, fmt="%d")
np.savetxt("y_pred.txt", y_pred_bool, fmt="%d")

print("Fine-tuning the model...")
model = load_model("lung_disease_model.keras")
model.compile(optimizer=AdamW(learning_rate=1e-5), loss='categorical_crossentropy', metrics=['accuracy'])

history_finetune = model.fit(
    train_dataset,
    epochs=30,
    validation_data=val_dataset,
    callbacks=[early_stopping, reduce_lr]
)

def plot_training_history(history):
    plt.figure(figsize=(12, 4))

    plt.subplot(1, 2, 1)
    plt.plot(history.history['accuracy'])
    plt.plot(history.history['val_accuracy'])
    plt.title('Model Accuracy')
    plt.ylabel('Accuracy')
    plt.xlabel('Epoch')
    plt.legend(['Train', 'Validation'], loc='upper left')

    plt.subplot(1, 2, 2)
    plt.plot(history.history['loss'])
    plt.plot(history.history['val_loss'])
    plt.title('Model Loss')
    plt.ylabel('Loss')
    plt.xlabel('Epoch')
    plt.legend(['Train', 'Validation'], loc='upper left')

    plt.tight_layout()
    plt.show()

print("Plotting training history...")
plot_training_history(history_finetune)

def plot_data_distribution(data_generator, title):
    class_counts = data_generator.classes
    class_labels = list(data_generator.class_indices.keys())

    plt.figure(figsize=(10, 5))
    plt.bar(class_labels, [np.sum(class_counts == i) for i in range(len(class_labels))])
    plt.xlabel("Classes")
    plt.ylabel("Number of Images")
    plt.title(f"Data Distribution - {title}")
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()

print("Plotting data distribution...")
plot_data_distribution(train_dataset, "Training Set")
plot_data_distribution(val_dataset, "Validation Set")
plot_data_distribution(test_dataset, "Test Set")

print("Analysis complete!")
```

```python
from google.colab import files
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
import io
import cv2
from tensorflow.keras.models import load_model
from tensorflow.keras.applications.densenet import preprocess_input
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.optimizers import Adam
import os
import pandas as pd
import seaborn as sns
from sklearn.metrics import confusion_matrix

# Ensure dataset_dir is correctly set in the global scope for functions in this cell
# This overwrites any potentially stale global variable with the correct path from the first cell.
# dataset_path is assumed to be available from the first cell's execution.
if 'dataset_path' in globals():
    temp_dataset_dir = None
    for root, dirs, files_in_dir in os.walk(dataset_path):
        if "train" in dirs and "val" in dirs and "test" in dirs:
            temp_dataset_dir = root
            break
    if temp_dataset_dir:
        dataset_dir = temp_dataset_dir
    else:
        print("Warning: Could not find 'train', 'val', 'test' directories under dataset_path.")
        # Fallback if the above fails, though it shouldn't if the first cell ran correctly
        dataset_dir = os.path.join(dataset_path, 'Lung Disease Dataset')
        if not os.path.exists(dataset_dir):
            print(f"Critical Error: Derived dataset_dir {dataset_dir} does not exist.")
else:
    print("Critical Error: 'dataset_path' not found in global scope. Please run the first cell.")
    dataset_dir = "/kaggle/input/lungs-disease-dataset-4-types/Lung Disease Dataset" # Fallback, should not be reached if first cell is run

# Use the dataset_dir determined during the initial download
# dataset_dir = "/kaggle/input/lungs-disease-dataset-4-types/Lung Disease Dataset" # Original hardcoded path

def predict_lung_disease(debug_mode=True, model_path="lung_disease_model.keras"):
    """
    Handles multiple image uploads and batch prediction with diagnostics.

    Args:
        debug_mode (bool): If True, displays additional diagnostic information.
        model_path (str): Path to the trained model.
    """
    print(f"Loading the trained model from {model_path}...")
    try:
        model = load_model(model_path)

        # Use the global dataset_dir variable from the previous cells
        global dataset_dir

        train_dir = os.path.join(dataset_dir, 'train')
        if not os.path.exists(train_dir):
            print("Error: Dataset directory does not exist. Check `dataset_dir` path.")
            return

        class_names = sorted(os.listdir(train_dir))
        print(f"Detected classes: {class_names}")

        if debug_mode:
            evaluate_model_performance(model, class_names)

        print("\n==== LUNG DISEASE PREDICTION TOOL ====")
        print("Upload one or more images of lung X-rays or CT scans (JPG, JPEG, PNG).")

        uploaded = files.upload()
        if not uploaded:
            print("No files uploaded. Please try again.")
            return

        print(f"\nTotal images uploaded: {len(uploaded)}")

        for filename, content in uploaded.items():
            print(f"\nProcessing image: {filename}")
            image = Image.open(io.BytesIO(content)).convert('RGB')
            image_array = np.array(image)
            original_height, original_width = image_array.shape[:2]
            print(f"Original image dimensions: {original_width}x{original_height}")

            image_resized = cv2.resize(image_array, (224, 224))

            plt.figure(figsize=(8, 6))
            plt.imshow(image_array)
            plt.title(f"Uploaded Image: {filename}")
            plt.axis('off')
            plt.show()

            if debug_mode:
                plt.figure(figsize=(8, 6))
                plt.imshow(image_resized)
                plt.title("Preprocessed Image (224x224)")
                plt.axis('off')
                plt.show()

            print("Analyzing image... Please wait.")

            image_expanded = np.expand_dims(image_resized, axis=0)
            image_preprocessed = preprocess_input(image_expanded)

            predictions = model.predict(image_preprocessed)[0]

            predicted_class_index = np.argmax(predictions)
            predicted_class = class_names[predicted_class_index]
            confidence = float(predictions[predicted_class_index]) * 100

            print("\n===== PREDICTION RESULTS =====")
            print(f"Image: {filename}")
            print(f"Diagnosis: {predicted_class}")
            print(f"Confidence: {confidence:.2f}%\n")

            for i, class_name in enumerate(class_names):
                print(f"{class_name}: {predictions[i]*100:.2f}%")

            print("=" * 50)

    except Exception as e:
        print(f"Error: {e}")
        import traceback
        traceback.print_exc()

def evaluate_model_performance(model, class_names):
    """
    Evaluates model performance on validation data.
    """
    try:
        val_datagen = ImageDataGenerator(preprocessing_function=preprocess_input)
        val_dataset = val_datagen.flow_from_directory(
            os.path.join(dataset_dir, 'val'),
            target_size=(224, 224),
            batch_size=32,
            class_mode='categorical',
            shuffle=False
        )

        print("Evaluating model on validation data...")
        val_loss, val_acc = model.evaluate(val_dataset)
        print(f"Validation accuracy: {val_acc:.4f}")
        print(f"Validation loss: {val_loss:.4f}")

        val_preds = model.predict(val_dataset)
        val_pred_classes = np.argmax(val_preds, axis=1)
        true_classes = val_dataset.classes

        cm = confusion_matrix(true_classes, val_pred_classes)
        cm_normalized = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]

        plt.figure(figsize=(10, 8))
        sns.heatmap(cm_normalized, annot=True, fmt='.2f', cmap='Blues',
                    xticklabels=class_names, yticklabels=class_names)
        plt.xlabel('Predicted')
        plt.ylabel('True')
        plt.title('Normalized Confusion Matrix')
        plt.show()

    except Exception as e:
        print(f"Error evaluating model: {e}")

def fix_biased_model():
    """
    Attempts to fix a model that is biased toward one class.
    """
    try:
        model_path = "lung_disease_model.keras"
        rebalanced_path = "lung_disease_model_rebalanced.keras"

        if not os.path.exists(model_path):
            print(f"Error: {model_path} not found. Train the model first.")
            return

        model = load_model(model_path)

        last_layer = model.layers[-1]
        weights, biases = last_layer.get_weights()

        new_biases = np.zeros_like(biases)

        threshold = np.mean(np.abs(weights)) + 2 * np.std(np.abs(weights))
        new_weights = np.clip(weights, -threshold, threshold)

        last_layer.set_weights([new_weights, new_biases])

        model.compile(loss='categorical_crossentropy',
                      optimizer=Adam(learning_rate=1e-5),
                      metrics=['accuracy'])

        model.save(rebalanced_path)
        print(f"Rebalanced model saved as '{rebalanced_path}'")

    except Exception as e:
        print(f"Error fixing model: {e}")

def explain_lung_diseases():
    """
    Provides information about different lung diseases in the dataset.
    """
    disease_info = {
        "COVID-19": "A respiratory illness caused by the SARS-CoV-2 virus.",
        "Normal": "Healthy lung tissue with no abnormalities.",
        "Pneumonia": "Infection causing inflammation in the lungs.",
        "Tuberculosis": "Bacterial infection affecting the lungs."
    }

    print("\n==== LUNG DISEASE INFORMATION ====")
    for disease, description in disease_info.items():
        print(f"\n🫁 {disease}: {description}")

    print("\n⚠️ This information is for educational purposes only. Consult a doctor for medical advice.")

def predict_with_rebalanced_model():
    """
    Uses the rebalanced model for prediction.
    """
    rebalanced_path = "lung_disease_model_rebalanced.keras"
    if os.path.exists(rebalanced_path):
        print(f"Using rebalanced model: {rebalanced_path}")
        predict_lung_disease(model_path=rebalanced_path)
    else:
        print("Rebalanced model not found. Run `fix_biased_model()` first.")

print("Updated lung disease prediction functions are now available.")
print("To predict images: predict_lung_disease()")
print("To fix model bias: fix_biased_model()")
print("To use the rebalanced model: predict_with_rebalanced_model()")
print("To learn about lung diseases: explain_lung_diseases()")
```
